```python

# 13. Car Price Prediction

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

# Features: car age (years), mileage (thousand km), engine size (L)
X = np.array([
    [1, 10, 2.0], [2, 20, 1.5], [3, 30, 1.8],
    [4, 40, 1.6], [5, 50, 2.0], [6, 60, 1.4],
    [2, 15, 2.5], [3, 25, 2.2], [1, 12, 1.8],
    [7, 80, 1.5], [4, 35, 2.4], [5, 55, 1.8],
    [8, 90, 1.6], [6, 65, 2.0], [2, 18, 2.0],
    [3, 28, 1.6], [7, 75, 2.2], [1, 8, 2.5],
    [4, 45, 1.8], [5, 48, 2.5]
])

# Illustrative prices in lakh rupees
y = np.array([
    12, 9, 8, 6.5, 6, 4.5, 14, 11, 13, 3.5,
    10, 7, 3, 6, 12, 9, 5, 16, 7, 9
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)
pred = model.predict(X_test)

print("Car Price Prediction")
print("Actual prices:", y_test)
print("Predicted prices:", np.round(pred, 2))
print("Mean Absolute Error:", round(mean_absolute_error(y_test, pred), 2))
print("R2 Score:", round(r2_score(y_test, pred), 2))

# Graph: Actual vs predicted prices
plt.scatter(y_test, pred, color="blue", label="Predictions")
low = min(y_test.min(), pred.min())
high = max(y_test.max(), pred.max())
plt.plot([low, high], [low, high], "r--", label="Perfect prediction")

plt.xlabel("Actual Price (Lakh Rs)")
plt.ylabel("Predicted Price (Lakh Rs)")
plt.title("Car Price Prediction")
plt.legend()
plt.grid(True)
plt.show()
```
