```python
# FIND-S Algorithm

import pandas as pd

data = pd.DataFrame({
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "AirTemp": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "High"],
    "Wind": ["Strong", "Strong", "Strong", "Strong"],
    "Water": ["Warm", "Warm", "Warm", "Cool"],
    "Forecast": ["Same", "Same", "Change", "Same"],
    "EnjoySport": ["Yes", "Yes", "No", "Yes"]
})

hypothesis = None

for _, row in data.iterrows():
    if row["EnjoySport"] == "Yes":
        example = list(row.iloc[:-1])

        if hypothesis is None:
            hypothesis = example
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != example[i]:
                    hypothesis[i] = "?"

        print("Current hypothesis:", hypothesis)

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
import matplotlib.pyplot as plt

steps = [1, 2, 3]
specific_attributes = [6, 5, 3]

plt.plot(steps, specific_attributes, marker="o")
plt.xlabel("Positive Example Number")
plt.ylabel("Specific Attribute Count")
plt.title("FIND-S Hypothesis Progress")
plt.xticks(steps)
plt.grid(True)
plt.show()
```
