```python

# Implementation of K-Nearest Neighbours (K-NN)
# with Graph Visualization

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier

# Step 1: Load the dataset
iris = load_iris()

# Use two features for easy graph visualization
X = iris.data[:, :2]
y = iris.target

# Step 2: Create the K-NN model
k = 5
model = KNeighborsClassifier(n_neighbors=k)

# Step 3: Train the model
model.fit(X, y)

# Step 4: Predict the class of a new sample
sample = [[5.1, 3.5]]
prediction = model.predict(sample)

print("K-NN Classification")
print("-------------------")
print("K value:", k)
print("Input sample:", sample[0])
print("Predicted class:", iris.target_names[prediction[0]])

# Step 5: Create a graph showing decision regions
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)
)

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Step 6: Plot the graph
plt.figure(figsize=(9, 6))

plt.contourf(xx, yy, Z, alpha=0.25, cmap="viridis")

for i, name in enumerate(iris.target_names):
    plt.scatter(
        X[y == i, 0],
        X[y == i, 1],
        label=name,
        edgecolor="black"
    )

# Mark the new sample
plt.scatter(
    sample[0][0],
    sample[0][1],
    color="red",
    marker="*",
    s=250,
    label="New sample"
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("K-Nearest Neighbours (K-NN) Classification")
plt.legend()
plt.grid(True)
plt.show()
```
