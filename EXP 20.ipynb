```python
# EXPERIMENT 20: FUTURE SALES PREDICTION
# Linear Regression

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Monthly sales data
months = np.arange(1, 25).reshape(-1, 1)

sales = np.array([
    120, 125, 130, 128, 140, 145,
    150, 148, 160, 165, 170, 175,
    180, 185, 190, 195, 205, 210,
    215, 220, 225, 230, 240, 245
])

model = LinearRegression()
model.fit(months, sales)

predicted_sales = model.predict(months)

print("Future Sales Prediction")
print("Mean Absolute Error:",
      mean_absolute_error(sales, predicted_sales))

print("R2 Score:", r2_score(sales, predicted_sales))

# Predict next 6 months
future_months = np.arange(25, 31).reshape(-1, 1)

future_sales = model.predict(future_months)

print("\nFuture Sales:")
for month, sale in zip(future_months.flatten(), future_sales):
    print("Month", month, "Sales:", round(sale, 2))

# Graph
plt.plot(months, sales, "o-", label="Actual Sales")
plt.plot(months, predicted_sales, "r--", label="Regression Line")
plt.plot(future_months, future_sales, "g*-", label="Future Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Future Sales Prediction using Linear Regression")
plt.legend()
plt.grid()
plt.show()
```
